# Resume Screening and Skill Matching System

### AI-Based Resume Screening using Sentence-BERT and FAISS

This project automatically processes resumes and compares them against a given job description.

### Workflow
1. Resume ingestion
2. Text extraction and preprocessing
3. Semantic embedding using Sentence-BERT
4. FAISS vector indexing
5. Job description matching
6. Candidate ranking
7. Screening decision
8. Excel report generation
